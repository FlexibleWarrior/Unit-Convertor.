```python
import streamlit as st
from streamlit_extras.stylable_container import stylable_container
from streamlit_extras.add_vertical_space import add_vertical_space

# Custom CSS for styling
st.set_page_config(
    page_title="Unit Converter",
    page_icon="📐",
    layout="centered",
    initial_sidebar_state="collapsed"
)

# Custom CSS
st.markdown("""
    <style>
        /* Main container styling */
        .main {
            max-width: 600px;
            padding: 2rem;
        }
        
        /* Input field styling */
        .stTextInput input {
            font-size: 24px;
            height: 60px;
            padding: 10px 15px;
        }
        
        /* Select box styling */
        .stSelectbox select {
            font-size: 16px;
            height: 60px;
        }
        
        /* Button styling */
        .stButton button {
            width: 100%;
            height: 50px;
            font-size: 16px;
            background-color: #4285F4;
            color: white;
        }
        
        /* Result styling */
        .result {
            font-size: 24px;
            font-weight: bold;
            margin-top: 20px;
            padding: 15px;
            background-color: #f8f9fa;
            border-radius: 8px;
            text-align: center;
        }
        
        /* Title styling */
        .title {
            font-size: 32px;
            text-align: center;
            margin-bottom: 30px;
            color: #4285F4;
        }
        
        /* Unit boxes styling */
        .unit-box {
            border: 1px solid #ddd;
            border-radius: 8px;
            padding: 15px;
            margin-bottom: 20px;
        }
    </style>
""", unsafe_allow_html=True)

# Unit conversion functions
def convert_length(value, from_unit, to_unit):
    # Convert to meters first
    to_meter = {
        'nanometer': 1e-9,
        'micrometer': 1e-6,
        'millimeter': 1e-3,
        'centimeter': 1e-2,
        'meter': 1,
        'kilometer': 1e3,
        'inch': 0.0254,
        'foot': 0.3048,
        'yard': 0.9144,
        'mile': 1609.34,
        'nautical mile': 1852
    }
    
    # Convert from meters to target unit
    from_meter = {k: 1/v for k, v in to_meter.items()}
    
    meters = value * to_meter[from_unit]
    return meters * from_meter[to_unit]

def convert_weight(value, from_unit, to_unit):
    # Convert to grams first
    to_gram = {
        'microgram': 1e-6,
        'milligram': 1e-3,
        'gram': 1,
        'kilogram': 1e3,
        'metric ton': 1e6,
        'ounce': 28.3495,
        'pound': 453.592,
        'stone': 6350.29,
        'US ton': 907185,
        'imperial ton': 1016050
    }
    
    # Convert from grams to target unit
    from_gram = {k: 1/v for k, v in to_gram.items()}
    
    grams = value * to_gram[from_unit]
    return grams * from_gram[to_unit]

def convert_temperature(value, from_unit, to_unit):
    # Convert to Celsius first
    if from_unit == 'celsius':
        celsius = value
    elif from_unit == 'fahrenheit':
        celsius = (value - 32) * 5/9
    elif from_unit == 'kelvin':
        celsius = value - 273.15
    
    # Convert from Celsius to target unit
    if to_unit == 'celsius':
        return celsius
    elif to_unit == 'fahrenheit':
        return celsius * 9/5 + 32
    elif to_unit == 'kelvin':
        return celsius + 273.15

def convert_area(value, from_unit, to_unit):
    # Convert to square meters first
    to_sqm = {
        'square millimeter': 1e-6,
        'square centimeter': 1e-4,
        'square meter': 1,
        'hectare': 1e4,
        'square kilometer': 1e6,
        'square inch': 0.00064516,
        'square foot': 0.092903,
        'square yard': 0.836127,
        'acre': 4046.86,
        'square mile': 2589988.11
    }
    
    # Convert from square meters to target unit
    from_sqm = {k: 1/v for k, v in to_sqm.items()}
    
    sqm = value * to_sqm[from_unit]
    return sqm * from_sqm[to_unit]

def convert_volume(value, from_unit, to_unit):
    # Convert to liters first
    to_liter = {
        'milliliter': 1e-3,
        'centiliter': 1e-2,
        'deciliter': 1e-1,
        'liter': 1,
        'kiloliter': 1e3,
        'cubic centimeter': 1e-3,
        'cubic meter': 1e3,
        'teaspoon (US)': 0.00492892,
        'tablespoon (US)': 0.0147868,
        'fluid ounce (US)': 0.0295735,
        'cup (US)': 0.236588,
        'pint (US)': 0.473176,
        'quart (US)': 0.946353,
        'gallon (US)': 3.78541,
        'cubic inch': 0.0163871,
        'cubic foot': 28.3168
    }
    
    # Convert from liters to target unit
    from_liter = {k: 1/v for k, v in to_liter.items()}
    
    liters = value * to_liter[from_unit]
    return liters * from_liter[to_unit]

def convert_time(value, from_unit, to_unit):
    # Convert to seconds first
    to_second = {
        'nanosecond': 1e-9,
        'microsecond': 1e-6,
        'millisecond': 1e-3,
        'second': 1,
        'minute': 60,
        'hour': 3600,
        'day': 86400,
        'week': 604800,
        'month': 2629746,
        'year': 31556952,
        'decade': 315569520,
        'century': 3155695200
    }
    
    # Convert from seconds to target unit
    from_second = {k: 1/v for k, v in to_second.items()}
    
    seconds = value * to_second[from_unit]
    return seconds * from_second[to_unit]

def convert_speed(value, from_unit, to_unit):
    # Convert to meters per second first
    to_mps = {
        'meters/second': 1,
        'kilometers/hour': 0.277778,
        'miles/hour': 0.44704,
        'knot': 0.514444,
        'feet/second': 0.3048,
        'speed of light': 299792458
    }
    
    # Convert from meters per second to target unit
    from_mps = {k: 1/v for k, v in to_mps.items()}
    
    mps = value * to_mps[from_unit]
    return mps * from_mps[to_unit]

def convert_pressure(value, from_unit, to_unit):
    # Convert to pascals first
    to_pascal = {
        'pascal': 1,
        'kilopascal': 1000,
        'megapascal': 1e6,
        'bar': 1e5,
        'psi': 6894.76,
        'atmosphere': 101325,
        'torr': 133.322,
        'mmHg': 133.322
    }
    
    # Convert from pascals to target unit
    from_pascal = {k: 1/v for k, v in to_pascal.items()}
    
    pascals = value * to_pascal[from_unit]
    return pascals * from_pascal[to_unit]

def convert_energy(value, from_unit, to_unit):
    # Convert to joules first
    to_joule = {
        'joule': 1,
        'kilojoule': 1000,
        'calorie': 4.184,
        'kilocalorie': 4184,
        'watt-hour': 3600,
        'kilowatt-hour': 3.6e6,
        'electronvolt': 1.60218e-19,
        'British thermal unit': 1055.06,
        'foot-pound': 1.35582
    }
    
    # Convert from joules to target unit
    from_joule = {k: 1/v for k, v in to_joule.items()}
    
    joules = value * to_joule[from_unit]
    return joules * from_joule[to_unit]

def convert_power(value, from_unit, to_unit):
    # Convert to watts first
    to_watt = {
        'watt': 1,
        'kilowatt': 1000,
        'megawatt': 1e6,
        'gigawatt': 1e9,
        'horsepower (metric)': 735.499,
        'horsepower (imperial)': 745.7,
        'foot-pound per second': 1.35582,
        'BTU per hour': 0.293071
    }
    
    # Convert from watts to target unit
    from_watt = {k: 1/v for k, v in to_watt.items()}
    
    watts = value * to_watt[from_unit]
    return watts * from_watt[to_unit]

def convert_data(value, from_unit, to_unit):
    # Convert to bytes first
    to_byte = {
        'bit': 0.125,
        'byte': 1,
        'kilobyte': 1024,
        'megabyte': 1024**2,
        'gigabyte': 1024**3,
        'terabyte': 1024**4,
        'petabyte': 1024**5,
        'kibibyte': 1024,
        'mebibyte': 1024**2,
        'gibibyte': 1024**3,
        'tebibyte': 1024**4,
        'pebibyte': 1024**5
    }
    
    # Convert from bytes to target unit
    from_byte = {k: 1/v for k, v in to_byte.items()}
    
    bytes_val = value * to_byte[from_unit]
    return bytes_val * from_byte[to_unit]

# Main app
def main():
    st.markdown('<h1 class="title">Unit Converter</h1>', unsafe_allow_html=True)
    
    # Category selection
    categories = {
        "Length": ["nanometer", "micrometer", "millimeter", "centimeter", "meter", 
                  "kilometer", "inch", "foot", "yard", "mile", "nautical mile"],
        "Weight": ["microgram", "milligram", "gram", "kilogram", "metric ton", 
                  "ounce", "pound", "stone", "US ton", "imperial ton"],
        "Temperature": ["celsius", "fahrenheit", "kelvin"],
        "Area": ["square millimeter", "square centimeter", "square meter", "hectare", 
                "square kilometer", "square inch", "square foot", "square yard", 
                "acre", "square mile"],
        "Volume": ["milliliter", "centiliter", "deciliter", "liter", "kiloliter", 
                  "cubic centimeter", "cubic meter", "teaspoon (US)", "tablespoon (US)", 
                  "fluid ounce (US)", "cup (US)", "pint (US)", "quart (US)", 
                  "gallon (US)", "cubic inch", "cubic foot"],
        "Time": ["nanosecond", "microsecond", "millisecond", "second", "minute", 
                "hour", "day", "week", "month", "year", "decade", "century"],
        "Speed": ["meters/second", "kilometers/hour", "miles/hour", "knot", 
                "feet/second", "speed of light"],
        "Pressure": ["pascal", "kilopascal", "megapascal", "bar", "psi", 
                   "atmosphere", "torr", "mmHg"],
        "Energy": ["joule", "kilojoule", "calorie", "kilocalorie", "watt-hour", 
                  "kilowatt-hour", "electronvolt", "British thermal unit", "foot-pound"],
        "Power": ["watt", "kilowatt", "megawatt", "gigawatt", "horsepower (metric)", 
                "horsepower (imperial)", "foot-pound per second", "BTU per hour"],
        "Data": ["bit", "byte", "kilobyte", "megabyte", "gigabyte", "terabyte", 
                "petabyte", "kibibyte", "mebibyte", "gibibyte", "tebibyte", "pebibyte"]
    }
    
    selected_category = st.selectbox("Select category", list(categories.keys()))
    
    # Input section
    col1, col2, col3 = st.columns([3, 2, 2])
    
    with col1:
        input_value = st.text_input("Enter value", value="1", key="input_value")
    
    with col2:
        from_unit = st.selectbox("From", categories[selected_category], key="from_unit")
    
    with col3:
        to_unit = st.selectbox("To", categories[selected_category], key="to_unit")
    
    # Convert button
    convert_clicked = st.button("Convert", type="primary")
    
    # Conversion logic
    result = None
    if convert_clicked and input_value:
        try:
            value = float(input_value)
            
            if selected_category == "Length":
                result = convert_length(value, from_unit, to_unit)
            elif selected_category == "Weight":
                result = convert_weight(value, from_unit, to_unit)
            elif selected_category == "Temperature":
                result = convert_temperature(value, from_unit, to_unit)
            elif selected_category == "Area":
                result = convert_area(value, from_unit, to_unit)
            elif selected_category == "Volume":
                result = convert_volume(value, from_unit, to_unit)
            elif selected_category == "Time":
                result = convert_time(value, from_unit, to_unit)
            elif selected_category == "Speed":
                result = convert_speed(value, from_unit, to_unit)
            elif selected_category == "Pressure":
                result = convert_pressure(value, from_unit, to_unit)
            elif selected_category == "Energy":
                result = convert_energy(value, from_unit, to_unit)
            elif selected_category == "Power":
                result = convert_power(value, from_unit, to_unit)
            elif selected_category == "Data":
                result = convert_data(value, from_unit, to_unit)
            
            if result is not None:
                # Format the result nicely
                if abs(result) >= 1e6 or abs(result) <= 1e-4:
                    formatted_result = f"{result:.4e}"
                else:
                    formatted_result = f"{result:,.4f}".rstrip('0').rstrip('.') if '.' in f"{result:,.4f}" else f"{result:,.4f}"
                
                # Display result
                st.markdown(f"""
                    <div class="result">
                        {value} {from_unit} = {formatted_result} {to_unit}
                    </div>
                """, unsafe_allow_html=True)
                
        except ValueError:
            st.error("Please enter a valid number")
    
    # Recent conversions (using session state)
    if 'history' not in st.session_state:
        st.session_state.history = []
    
    if result is not None and convert_clicked:
        conversion_text = f"{input_value} {from_unit} → {result:.4f} {to_unit}"
        st.session_state.history.append(conversion_text)
        
        # Keep only the last 5 conversions
        if len(st.session_state.history) > 5:
            st.session_state.history = st.session_state.history[-5:]
    
    # Display history
    if st.session_state.history:
        st.subheader("Recent Conversions")
        for item in reversed(st.session_state.history):
            st.markdown(f"🔹 {item}")

    # Footer
    add_vertical_space(2)
    st.markdown("""
    <div style="text-align: center; color: #666; font-size: 14px;">
        <p>Unit Converter - Inspired by Google's Unit Converter</p>
        <p>All conversions are approximate</p>
    </div>
    """, unsafe_allow_html=True)

if __name__ == "__main__":
    main()
    
```
